## 13.1 — Write the current date to today.txt

```python
from datetime import date

today = date.today()

with open("today.txt", "w") as file:
    file.write(today.isoformat())

print("Date saved:", today.isoformat())
```

## 13.2 — Read today.txt into today_string

```python
with open("today.txt", "r") as file:
    today_string = file.read()

print("today_string:", today_string)
print("Type:", type(today_string))
```

## 13.3 — Parse the date from today_string

```python
parsed_date = date.fromisoformat(today_string.strip())

print("Parsed date:", parsed_date)
print("Type:", type(parsed_date))
print("Year:", parsed_date.year)
print("Month:", parsed_date.month)
print("Day:", parsed_date.day)
```

## 15.1 — Create three separate processes

Each process waits a random number of seconds between zero and one,
prints the current time, and exits.

```python
%%writefile multiprocessing_example.py
import multiprocessing
import random
import time
from datetime import datetime


def show_time(process_number):
    delay = random.uniform(0, 1)
    time.sleep(delay)

    print(
        f"Process {process_number}: "
        f"current time: {datetime.now().strftime('%H:%M:%S.%f')}",
        flush=True
    )


if __name__ == "__main__":
    processes = []

    for number in range(1, 4):
        process = multiprocessing.Process(
            target=show_time,
            args=(number,)
        )
        processes.append(process)
        process.start()

    for process in processes:
        process.join()
```

```python
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "multiprocessing_example.py"],
    capture_output=True,
    text=True,
    check=True
)

print(result.stdout)
```
